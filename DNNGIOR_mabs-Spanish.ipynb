# Gapfilling con DNNGIOR usando una Red Neural taxonómica-específica

Por defecto, DNNgior utiliza una red neuronal integrada. Sin embargo, ocasionalmente, usted trabajará con bacterias complejas con requisitos de crecimiento específicos, como las Actinobacterias.

*Mycobacterium abscessus* (o *Mycobacteroides abscessus*, una larga história!) es una micobacteria altamente resistente a los antibióticos que causa infecciones respiratorias, de piel y mucosas muy severas en seres humanos. Las micobacterias son consideradas de dificil crecimiento y usualmente requieren medios de cultivo específicos para crecer en el laboratório. 

En este ejercicio, vamos a realizar gapfill de un modelo metabólico de *M. abscessus* con una  red neuronal específica entrenada con especies del Phylum Actinobacteria. 

Primero, cargue e inspeccione el modelo de *M. abscessus* con Cobra (archivo Mabs.sbml)

```python
import cobra
from cobra.io import read_sbml_model
draft_reconstruction = read_sbml_model('/home/carol/Desktop/IA/MeIA/Mabs.sbml')
```

```python
draft_reconstruction
```

```python
draft_reconstruction.summary()
```

```python
draft_reconstruction.optimize()
```

```python
draft_reconstruction.medium
```

Vamos a hacer primero el gapfill con la red neural integrada de DNNgior.

```python
import os, sys
path_to_mabs_model = ('/home/carol/Desktop/IA/MeIA/Mabs.sbml')
```

```python
import dnngior
gapfilled_model_complete = dnngior.Gapfill(draftModel = path_to_mabs_model, 
                                          medium = None, 
                                          objectiveName = 'bio1')
```

```python
gf_model_compl_med = gapfilled_model_complete.gapfilledModel.copy()
```

```python
gf_model_compl_med.optimize()
```

<div class="alert alert-block alert-info">
Note que antes de realizar el gapfilling el crecimiento era de 185.74 mmol/gDW/hr, después del gapfill el crecimiento es de 252.767 mmol/gDW/hr.

Ahora veamos cuantas reacciones adicionó

```python
print("Number of reactions added:", len(gapfilled_model_complete.added_reactions))
print("~~")
for reaction in gapfilled_model_complete.added_reactions:
    print(gf_model_compl_med.reactions.get_by_id(reaction).name)
```

# **Gapfilling con la red neural específica para Actinobacterias**

```python
gapfilled_model_complete_NN_ac = dnngior.Gapfill(draftModel = path_to_mabs_model, trainedNNPath= '/home/carol/Desktop/IA/TUTORIAL_X_MEETING/Phylum_Actinobacteria.npz',
                                          medium = None, 
                                          objectiveName = 'bio1')
```

```python
gf_model_compl_NN_ac = gapfilled_model_complete_NN_ac.gapfilledModel.copy()
```

```python
gf_model_compl_NN_ac.optimize()
```

<div class="alert alert-block alert-info">
Con la red neural regular el crecimiento era de 252.767 mmol/gDW/hr, con la red neural específica el crecimiento aumentó a 285.976 mmol/gDW/hr!

Veamos cuantas reacciones fueron adicionadas

```python
print("Number of reactions added:", len(gapfilled_model_complete_NN_ac.added_reactions))
print("~~")
for reaction in gapfilled_model_complete_NN_ac.added_reactions:
    print(gf_model_compl_NN_ac.reactions.get_by_id(reaction).name)
```

Con la red neural regular 16 reaccionesfueron adicionadas a nuestro modelo, pero con la red específica para Actinobacterias 4 nuevas reacciones fueron adicionadas.
